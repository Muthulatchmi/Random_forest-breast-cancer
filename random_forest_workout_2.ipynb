```python
x=df.drop(["target","diagnosis"],axis=1)
y=df["target"]
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.25,random_state=33)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler

preprocessor = ColumnTransformer(
    transformers=[
        ('scaler', StandardScaler(), x.columns.tolist())
    ],
    remainder='drop'
)
```

```python
processed_x_train = preprocessor.fit_transform(x_train)
processed_x_test = preprocessor.transform(x_test)
```

```python
processed_x_train[0]
```

```python
from sklearn.utils.class_weight import compute_sample_weight
sample_weights = compute_sample_weight(class_weight={0:1,1:5},y=y_train)
```

```python
from imblearn.over_sampling import SMOTE
smote=SMOTE()
```

```python
x_train_smote,y_train_smote=smote.fit_resample(processed_x_train,y_train)
```

```python
from sklearn.ensemble import RandomForestClassifier
rf_clf = RandomForestClassifier()
rf_clf.fit(processed_x_train,y_train)
```

```python
y_predict=rf_clf.predict(processed_x_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_predict)
```

```python
from sklearn.model_selection import RandomizedSearchCV
from sklearn.ensemble import RandomForestClassifier

rf_clf1 = RandomForestClassifier(random_state=42)
random_grid={
    'n_estimators': [100,200,300],
    'max_features': [ 'sqrt', 'log2'],
    'max_depth' : [10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4 ]
}

rf_cv = RandomizedSearchCV(
    estimator=rf_clf1,
    param_distributions=random_grid,
    n_iter=20,              # KEY LIMIT
    scoring='accuracy',     # faster
    cv=3,
    verbose=2,
    n_jobs=2,               # DO NOT USE -1 FOR STUDENTS
    random_state=42
)

rf_cv.fit(processed_x_train, y_train)
```

```python
rf_cv.best_params_
```

```python
rf_clf2 = RandomForestClassifier(n_estimators=100,min_samples_leaf=2,min_samples_split=5,max_features="sqrt",max_depth=20)
rf_clf2.fit(processed_x_train, y_train, sample_weight=sample_weights)
```

```python
from sklearn.metrics import accuracy_score
y_predict=rf_clf2.predict(processed_x_test)    # testing
accur=accuracy_score(y_test,y_predict)   # checking performance
accur
```
